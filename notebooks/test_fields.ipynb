```python
# This script sets up a step-index fibre with specific parameters.
fibre = fib.StepIndexFibre(
    core_radius = 250*u.nm, 
    eps_core = 4,    
    eps_clad = 1,    
    mu_core  = 1,    
    mu_clad  = 1   
) 
```

```python
mode = fibre.HE(ell=1, n=1, wl=500*u.nm, a_plus=1, a_minus=0)
display(mode)
```
